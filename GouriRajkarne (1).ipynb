Chunking

```python
def chunk_text(text, chunk_size=500):
    chunks = []

    for i in range(0, len(text), chunk_size):
        chunks.append(text[i:i+chunk_size])

    return chunks

chunks = chunk_text(text)
```

Ranking

```python
from rank_bm25 import BM25Okapi

tokenized_chunks = [
    chunk.lower().split()
    for chunk in chunks
]

bm25 = BM25Okapi(tokenized_chunks)
```


LaBSE :	Cross-language semantic embeddings

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    'sentence-transformers/LaBSE'
)

embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True
)
```

Embeddings

```python
import faiss
import numpy as np

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(np.array(embeddings))
```

Hybrid Search

```python
def hybrid_search(query, top_k=5):

    # BM25 retrieval
    tokenized_query = query.lower().split()

    bm25_scores = bm25.get_scores(tokenized_query)

    bm25_top = np.argsort(bm25_scores)[::-1][:top_k]

    # Dense retrieval
    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    )

    D, I = index.search(query_embedding, top_k)

    dense_top = I[0]

    # Merge results
    combined = list(set(bm25_top.tolist() + dense_top.tolist()))

    return [chunks[i] for i in combined]
```

Cross Encoder

```python
from sentence_transformers import CrossEncoder
import numpy as np

reranker = CrossEncoder(
    'BAAI/bge-reranker-v2-m3'
)

def rerank(query, retrieved_chunks):

    pairs = [
        [query, chunk]
        for chunk in retrieved_chunks
    ]

    scores = reranker.predict(pairs)

    normalized_scores = 1 / (1 + np.exp(-scores))

    ranked = sorted(
        zip(retrieved_chunks, normalized_scores),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked
```

```python
queries = [
    "पॉलिसी दुर्घटना में क्या कवर करती है?",
    "Cette police couvre-t-elle le vol ?",
    "Deckt die Police Motorschäden ab?"
]

for query in queries:

    print("\n==============================")
    print("QUERY:", query)
    print("==============================")

    # 1: Hybrid Retrieval
    retrieved = hybrid_search(query)

    # 2: Reranking
    ranked_results = rerank(query, retrieved)

    # 3: Show Top Results
    for idx, (chunk, score) in enumerate(ranked_results[:3]):

        print(f"\nTop Result {idx+1}")
        print("Relevance Score:", round(score, 3))
        print(chunk[:1000])
```

TASK 4 — AI Evaluation Agent


Faithfulness :	Is answer supported by retrieved policy text?

Relevancy :	Does answer match query intent?

Context Precision :	Did retrieval fetch correct clauses?

Groundedness :	Are statements backed by context?

Latency	: Full pipeline speed

```python
import time
import numpy as np
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer('sentence-transformers/LaBSE')
```

```python
#Faithfulness Score
def faithfulness_score(answer, context):

    emb_a = embedding_model.encode([answer])
    emb_c = embedding_model.encode([context])

    score = np.dot(emb_a, emb_c.T)[0][0]

    return float(score)

# Relevancy Score
def relevancy_score(query, answer):

    emb_q = embedding_model.encode([query])
    emb_a = embedding_model.encode([answer])

    score = np.dot(emb_q, emb_a.T)[0][0]

    return float(score)

#Context Precision
def context_precision(query, retrieved_chunks):

    q_emb = embedding_model.encode([query])

    chunk_embs = embedding_model.encode(retrieved_chunks)

    scores = np.dot(chunk_embs, q_emb.T).flatten()

    return float(np.mean(scores))

#Groundness Check
def groundedness_check(answer, context):

    return all(
        sentence.lower() in context.lower()
        for sentence in answer.split(".")
        if len(sentence.strip()) > 5
    )

#Latency Check
def track_latency(start_time):

    return round(time.time() - start_time, 3)
```

```python
def evaluate_rag(query, answer, retrieved_chunks, start_time):

    context = " ".join(retrieved_chunks)

    return {
        "faithfulness": faithfulness_score(answer, context),
        "relevancy": relevancy_score(query, answer),
        "context_precision": context_precision(query, retrieved_chunks),
        "groundedness": groundedness_check(answer, context),
        "latency_sec": track_latency(start_time)
    }
```

```python
import numpy as np

def detect_hallucination(metrics):

    faithfulness = metrics.get("faithfulness", 0)
    grounded = metrics.get("groundedness", False)

    # Decision logic
    if faithfulness < 0.6 or grounded == False:
        return True

    return False
```

```python
import time
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="google/flan-t5-base"
)

def generate_answer(query, context):

    prompt = f"""
    Answer the question using only the context.

    Question:
    {query}

    Context:
    {context}

    Answer:
    """

    output = generator(prompt, max_new_tokens=120)

    return output[0]["generated_text"]
```

```python
import time

query = "What loss and damage are covered under the insurance?"

start = time.time()

# 1. Retrieval
retrieved = hybrid_search(query)
ranked = rerank(query, retrieved)

top_context = ranked[0][0]

# 2. LLM GENERATION
answer = generate_answer(query, top_context)

print("\n===== GENERATED ANSWER =====")
print(answer)

# 3. Evaluation
metrics = evaluate_rag(query, answer, [top_context], start)

print("\n===== METRICS =====")
print(metrics)

# 4. Hallucination Check
is_hallucination = detect_hallucination(metrics)

print("\n==============================")
print("HALLUCINATION CHECK")
print("==============================")

if is_hallucination:
    print("WARNING: Hallucinated or weak response detected")
else:
    print("Response is grounded and safe")
```
